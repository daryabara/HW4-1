```python
import pandas as pd
import numpy
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from catboost import CatBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier, AdaBoostClassifier, ExtraTreesClassifier
import nltk
import catboost
import matplotlib
import seaborn
```

```python
with open("dataset_homework.csv", encoding='latin-1') as f:
    lines = f.readlines()

parsed_dataset = []

for line in lines:
    fields = line.split(",")
    fields = [field.strip().strip('"').strip() for field in fields]
    fields[0], fields[1] = int(fields[0]), int(fields[1])
    parsed_dataset.append(fields)

count_incorrect = 0

final_dataset = []

for el in parsed_dataset:
    if len(el) == 6:
        final_dataset.append(el)

data = pd.DataFrame(final_dataset, columns=["sentiment", "id", "date", "flag", "user", "text"])  # Подставь свои названия колонок
data.head()
```

```python
# просмотр информации информацию о датасете
print("\nИнформация о данных:")
print(data.info(1000))

# Проверим сколько в датасете положительных и отрицтельных твитов
print("\nРаспределение тональностей:")
print(data["sentiment"].value_counts())

# в исходных данных отрицательные твиты имеют обозначение 0, положительные 4, для удобства преобразуем 4 в 1
print("\nПримеры текстов:")
print("Отрицательный твит (0):", data[data["sentiment"] == 0]["text"].iloc[0])
data["sentiment"] = data["sentiment"].replace({4: 1})
print("Положительный твит (4 → 1):", data[data["sentiment"] == 1]["text"].iloc[0])


# Проверим изменения
print("\nПосле преобразования меток:")
print(data["sentiment"].value_counts())

# выведем первые 20 твитов с тональностью
print("\nПервые 20 твитов с тональностью:")
for i in range(20):
    sentiment = "Положительный" if data["sentiment"].iloc[i] == 1 else "Отрицательный"
    print(f"{i+1}. [{sentiment}] {data['text'].iloc[i]}")
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Загрузка необходимых ресурсов NLTK
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')  # Для токенизатора

def clean_text(text):
    # Приводим к строковому типу на случай NaN
    text = str(text)

    # Удаление ссылок (полная версия regex для URL)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Удаление хэштегов и упоминаний
    text = re.sub(r'[@#]\w+', '', text)

    # Удаление HTML-сущностей (если есть)
    text = re.sub(r'&[a-z]+;', '', text)

    # 4. Удаление спецсимволов (кроме апострофа для английских сокращений)
    text = re.sub(r"[^a-zA-Z'\s]", '', text)

    # 5. Приведение к нижнему регистру
    text = text.lower()

    try:
        # 6. Токенизация с обработкой возможных ошибок
        tokens = word_tokenize(text, language='english')
    except:
        # Фолбэк на простой сплит при ошибках токенизации
        tokens = text.split()

    # Удаление стоп-слов с обработкой апострофов
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]

    # Лемматизация с учетом частых ошибок
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens if len(word) > 1]  # Игнорируем одиночные символы

    return ' '.join(tokens)

# Применяем очистку к тексту твитов
data['cleaned_text'] = data['text'].apply(clean_text)

# Векторизаторы с обработкой ошибок
try:
    # Bag of Words с обработкой ошибок
    bow_vectorizer = CountVectorizer(
        max_features=5000,
        ngram_range=(1, 1),
        token_pattern=r'\b\w+\b',  # Четкий паттерн для токенов
        strip_accents='unicode'
    )
    X_bow = bow_vectorizer.fit_transform(data['cleaned_text'])

    # N-граммы с обработкой ошибок
    ngram_vectorizer = CountVectorizer(
        max_features=5000,
        ngram_range=(2, 3),
        token_pattern=r'\b\w+\b',
        min_df=2  # Игнорировать редкие n-граммы
    )
    X_ngram = ngram_vectorizer.fit_transform(data['cleaned_text'])

    # TF-IDF с обработкой ошибок
    tfidf_vectorizer = TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 3),
        token_pattern=r'\b\w+\b',
        sublinear_tf=True  # Сглаживание значений TF
    )
    X_tfidf = tfidf_vectorizer.fit_transform(data['cleaned_text'])

except Exception as e:
    print(f"Ошибка векторизации: {e}")
    # Альтернатива с простой векторизацией
    bow_vectorizer = CountVectorizer(max_features=5000)
    X_bow = bow_vectorizer.fit_transform(data['cleaned_text'])
    X_ngram = X_bow
    X_tfidf = X_bow

# Проверка результатов
print("Пример очищенного текста:", data['cleaned_text'].iloc[0])
print("\nBag of Words features:", bow_vectorizer.get_feature_names_out()[:10])
print("N-gram features:", ngram_vectorizer.get_feature_names_out()[:10])
print("TF-IDF features:", tfidf_vectorizer.get_feature_names_out()[:10])

# Сохранение данных
data[['cleaned_text', 'sentiment']].to_csv('cleaned_tweets.csv', index=False)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
#смотрим на данные после манипуляций
print(data.head(10))
print(data['sentiment'].value_counts())

# Создаём TF-IDF векторизатор (берём 5000 самых важных слов), анализируем 5000 самых частых слов из датасета
tfidf = TfidfVectorizer(max_features=5000)
X = tfidf.fit_transform(data["cleaned_text"])
y = data["sentiment"]

# Разделяем данные на обучающую и тестовую выборки 70 на 30
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

#примеры плохих и хороших слов
feature_names = tfidf.get_feature_names_out()
print("топ 100 слов", feature_names[:100])
```

```python
from sklearn.ensemble import GradientBoostingClassifier, AdaBoostClassifier
from catboost import CatBoostClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
#будем тестировать 5 моделей, инициализируем модели
models = {
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "CatBoost": CatBoostClassifier(verbose=0, random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Naive Bayes": MultinomialNB()
}
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
import pandas as pd

# Инициализация списка для хранения результатов
results = []

# Перебор моделей
for name, model in models.items():
    print(f"\nОбучение {name}...")

    try:
        # Обучение модели
        model.fit(X_train.toarray(), y_train)

        # Предсказание
        y_pred = model.predict(X_test.toarray())

        # Расчет метрик
        accuracy = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred)
        recall = recall_score(y_test, y_pred)

        # Сохранение результатов
        results.append({
            "Model": name,
            "Accuracy": round(accuracy, 4),
            "F1": round(f1, 4),
            "Precision": round(precision, 4),
            "Recall": round(recall, 4)
        })

        print(f"Метрики для {name}:")
        print(f"Accuracy: {accuracy:.4f}, F1: {f1:.4f}, Precision: {precision:.4f}, Recall: {recall:.4f}")

    except Exception as e:
        print(f"Ошибка при обучении {name}: {str(e)}")
        results.append({
            "Model": name,
            "Accuracy": 0,
            "F1": 0,
            "Precision": 0,
            "Recall": 0
        })

# Создаем DataFrame с результатами
if results:  # Проверяем, что есть результаты
    results_df = pd.DataFrame(results).sort_values(by="F1", ascending=False).reset_index(drop=True)
    display(results_df)

    # Вывод лучшей модели
    best_model = results_df.iloc[0]
    print(f"\nЛучшая модель: {best_model['Model']} с F1-score: {best_model['F1']:.4f}")

    # Визуализация сравнения моделей
    metrics = ['Accuracy', 'F1', 'Precision', 'Recall']
    results_df.set_index('Model')[metrics].plot(kind='bar', figsize=(12, 6))
    plt.title('Сравнение моделей по метрикам')
    plt.ylabel('Значение метрики')
    plt.xticks(rotation=45)
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.show()
else:
    print("Нет результатов для отображения")
```
